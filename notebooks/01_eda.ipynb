# Provider fraud dataset exploration

This notebook inspects the supplied source tables and provider-level target distribution. It avoids displaying beneficiary or claim-level records.

```python
from pathlib import Path
import sys

for candidate in (Path.cwd(), *Path.cwd().parents):
    if (candidate / "src" / "fraud_detection").is_dir():
        PROJECT_ROOT = candidate
        break
else:
    raise RuntimeError("Open this notebook from within the project directory.")

sys.path.insert(0, str(PROJECT_ROOT / "src"))

from fraud_detection.data import load_training_tables

tables = load_training_tables()
for name, frame in tables.items():
    print(f"{name}: {frame.shape[0]:,} rows x {frame.shape[1]:,} columns")
```

```python
labels = tables["provider_labels"]
label_counts = labels["PotentialFraud"].value_counts(dropna=False)
label_counts
```

```python
import matplotlib.pyplot as plt

ax = label_counts.plot(kind="bar", color=["steelblue", "crimson"])
ax.set_title("Provider labels in the training dataset")
ax.set_xlabel("PotentialFraud label")
ax.set_ylabel("Provider count")
plt.tight_layout()
plt.show()
```
